# Part 1: Lists

Every variable you made last session held exactly one value. Real data doesn't work that way, you'll have a list of sites, or a column of completion rates. A **list** is Python's way of holding many values in one variable.

You build one with square brackets, items separated by commas:

```python
sites = ["Washington D.C.", "New York", "Pennsylvania"]
print(sites)
```

## Indexing: getting one item out of a list

Each item has a position, called an **index**. Python starts counting at 0, not 1. This trips up literally everyone at first, so get used to it now.

| Item | "Washington D.C." | "New York" | "Pennsylvania" |
|---|---|---|---|
| Index (from the front) | `0` | `1` | `2` |
| Index (from the back) | `-3` | `-2` | `-1` |

```python
print(sites[0])   # first item
print(sites[2])   # third item
print(sites[-1])  # last item, counting from the end
```

## Length, and the error everyone hits

`len()` tells you how many items are in a list. This next one **fails on purpose**, run it and read the last line of the error:

```python
sites = ["Washington D.C.", "New York", "Pennsylvania"]
print(len(sites))
print(sites[3])   # there is no index 3 in a 3-item list
```

**What happened:** an `IndexError`. There are 3 items, sitting at indices 0, 1, and 2. Index 3 doesn't exist. Same rule as last session's type errors: read the last line, don't panic, fix the index.

## Adding to a list

Lists aren't fixed once created. `.append()` adds an item to the end:

```python
sites.append("Virginia")
print(sites)
print(len(sites))
```

# Part 2: Loops

Printing each item in a list one line at a time doesn't scale, imagine doing that for 50 sites. A **for loop** runs the same block of code once for every item in a list, automatically.

```python
sites = ["Washington D.C.", "New York", "Pennsylvania"]

for site in sites:
    print(site)
```

Read this as "for each `site` in `sites`, do the following." `site` is a variable name you're choosing, it holds whatever item the loop is currently on. `sites` is the list from before.

**Indentation is not decoration.** The indented line(s) under `for site in sites:` are the loop body, everything Python repeats. Get the indentation wrong and either nothing runs or Python throws an error. Colab auto-indents for you after the colon; don't fight it.

## The accumulator pattern

A common thing to do in a loop is build up a running total. Start a variable at zero *before* the loop, then update it *inside* the loop:

```python
completion_rates = [72.5, 88.0, 65.3, 91.2]

total = 0
for rate in completion_rates:
    total = total + rate

average = total / len(completion_rates)
print(f"Average completion rate: {round(average, 1)}%")
```

This pattern, start a counter outside the loop and update it inside, shows up constantly: running totals, counts, building a new list. Get comfortable with it now; you'll use it for the rest of the series.

# Part 3: Conditionals

Everything so far runs the same way no matter what the data looks like. A **conditional** lets your code branch: do one thing if something is true, another thing if it isn't.

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

**Careful:** `=` assigns a value (`score = 10`); `==` checks if two things are equal (`score == 10`). Mixing these up is the second most common beginner error, right after the index-0 thing.

```python
rate = 91.2

if rate >= 90:
    print("On track")
else:
    print("Needs follow-up")
```

`if` checks a condition. If it's `True`, the indented block underneath runs. `else` catches everything that isn't. Chain more conditions with `elif` ("else if"):

```python
rate = 65.3

if rate >= 90:
    status = "On track"
elif rate >= 75:
    status = "Monitor"
else:
    status = "Needs follow-up"

print(f"{rate}% -> {status}")
```
