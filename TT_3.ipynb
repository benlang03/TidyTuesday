# Part 1: Dictionaries

## The problem with parallel lists

With only lists, connecting sites to rates means keeping two lists, `sites` and `rates`, where New York sits at index 1 in *both*. That works until someone sorts one list and not the other, or deletes a site from one and forgets the other. Then every number is silently attached to the wrong site, and nothing errors out to warn you. That's the worst kind of bug.

## Building a dictionary

A **dictionary** stores **key: value** pairs. Instead of looking things up by position (0, 1, 2), you look them up by a label you choose. You build one with curly braces:

```python
rates = {
    "Washington D.C.": 72.5,
    "New York": 88.0,
    "Pennsylvania": 65.3
}
print(rates)
```

The part before each colon is the **key** (the label), and the part after is the **value**. Keys are usually strings; values can be any type, including lists or even other dictionaries.

## Looking up a value

Same square brackets as a list, but you put the key inside instead of an index number:

```python
print(rates["New York"])
print(rates["Pennsylvania"])
```

No counting, no keeping two lists in sync. The rate lives right next to the name it belongs to.

## The error everyone hits (dictionary edition)

Last session it was `IndexError`. This one **fails on purpose**, run it and read the last line:

```python
print(rates["Virginia"])   # Virginia isn't in the dictionary
```

**What happened:** a `KeyError`. You asked for a key that doesn't exist. Keys also have to match *exactly*: `"new york"` and `"New York"` are different keys, so capitalization and extra spaces count.

If you're not sure a key is there, `rates.get("Virginia")` returns `None` (Python's "nothing") instead of crashing.

## Adding and updating

The same syntax does both. If the key exists, its value gets replaced. If it doesn't, a new pair is added:

```python
rates["Virginia"] = 80.1    # new key -> added
rates["New York"] = 90.5    # existing key -> updated
print(rates)
print(len(rates))           # len() counts key: value pairs
```

**Careful:** updating and adding look identical, so a typo in a key name won't error, it creates a second entry. `rates["New Yrok"] = 95` adds a new site instead of fixing New York. When a dictionary has one more item than you expected, this is usually why.

## Looping over a dictionary

`.items()` hands you each key and value together, one pair per pass through the loop. You name both variables:

```python
for site, rate in rates.items():
    print(f"{site}: {rate}%")
```

Read it as "for each `site` and `rate` pair in `rates`."

# Part 2: Functions

You've already been *using* functions since Session 1: `print()`, `len()`, `round()`, `type()`. Someone wrote the code once, gave it a name, and now everyone reuses it. Today you write your own.

## Why bother?

Look back at Session 2. The "On track / Monitor / Needs follow-up" logic had to be rewritten every time you wanted to check a rate. Copy-pasting code means that when the rule changes (say, "On track" moves from 90 to 85), you have to find and fix every copy, and you'll miss one.

A function lets you write the logic **once**:

```python
def get_status(rate):
    if rate >= 90:
        return "On track"
    elif rate >= 75:
        return "Monitor"
    else:
        return "Needs follow-up"
```

Running that cell doesn't print anything, and that's correct. It only *defines* the function. Nothing happens until you **call** it:

```python
print(get_status(91.2))
print(get_status(80.0))
print(get_status(65.3))
```

Breaking down the pieces:

| Piece | What it does |
|---|---|
| `def` | Tells Python "I'm defining a function" |
| `get_status` | The name you choose; you use it to call the function later |
| `(rate)` | The **parameter**: a placeholder for whatever value gets passed in |
| `:` and indentation | Same as loops and `if`: the indented block is the function body |
| `return` | Sends a value back to whoever called the function, and exits immediately |

When you call `get_status(91.2)`, Python sets `rate = 91.2` inside the function, runs the body, and hands back whatever gets `return`ed.

## `print` vs. `return`: the most important distinction today

These look similar in a notebook, because both make something show up on screen. They are not the same. This cell has a function that **prints** instead of **returns**. Run it and look closely at the second line of output:

```python
def double_print(x):
    print(x * 2)

def double_return(x):
    return x * 2

result_a = double_print(5)
print(f"result_a is: {result_a}")

result_b = double_return(5)
print(f"result_b is: {result_b}")
```

**What happened:** `double_print` showed `10` on screen, but `result_a` is `None`. `print()` only *displays* a value; it doesn't give anything back. `double_return` handed its value back, so `result_b` actually holds `10` and you can keep working with it.

The rule of thumb: **if you'll ever want to store, compare, or do math with a function's output, use `return`.** Printing is for you to look at; returning is for your code to use. Almost every function you write for real analysis should `return`.

# Part 3: Putting It Together, A List of Dictionaries

Here's the payoff. Picture a spreadsheet: every **row** is one site, and every **column** is a piece of information about it. In Python, you represent that as a **list** (the rows) of **dictionaries** (each row, with column names as keys):

```python
site_data = [
    {"site": "Washington D.C.", "enrolled": 45, "completed": 38},
    {"site": "New York",        "enrolled": 12, "completed": 7},
    {"site": "Pennsylvania",    "enrolled": 88, "completed": 81},
    {"site": "Virginia",        "enrolled": 30, "completed": 19},
]

print(site_data[0])             # the first row (a dictionary)
print(site_data[0]["site"])     # one "cell": first row, "site" column
```

Notice you're chaining both skills: `site_data[0]` uses a list **index** to get a row, then `["site"]` uses a dictionary **key** to get a column from that row.

Now combine it with a loop and a function, and you've got a small analysis pipeline:

```python
def completion_rate(enrolled, completed):
    return round(completed / enrolled * 100, 1)

for row in site_data:
    name = row["site"]
    rate = completion_rate(row["enrolled"], row["completed"])
    print(f"{name}: {rate}% -> {get_status(rate)}")
```

That's four lines of loop doing what would otherwise be a manual formula in every row of Excel. Add 500 sites and the code doesn't change at all.

# What you learned today

- Dictionaries store **key: value** pairs, so related data stays together instead of in separate lists that can drift out of sync
- Look up values with `dict["key"]`; a missing key raises a `KeyError`
- The same syntax adds a new key or updates an existing one, so typos create new entries silently
- `.items()` lets you loop over keys and values together
- `def` creates a function; parameters are its inputs, `return` sends back its output
- `print` displays a value, `return` hands it back; if you'll reuse the result, `return` it
- A **list of dictionaries** is Python's version of a spreadsheet: rows are dictionaries, columns are keys
